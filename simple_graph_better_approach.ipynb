```python
#importing modules and libraries
import networkx as nx

G=nx.DiGraph()

#adding nodes
nodes = ["A","B","C","D","E",]
G.add_nodes_from(nodes)

#adding edges
edges = [("A","B") , ("A","C") , ("B","D") ,( "C","D") , ("D","E") , ("B","E")]
G.add_edges_from(edges)

#adding position of nodes
pos={
    "A" : (2 , 4),
    "B" : (5 , 5.5),
    "C" : (3.5 , 2),
    "D" : (6 , 3),
    "E" : (8 , 3.5)
}

#visualaizing the graph
nx.draw(
    G,
    with_labels = True ,
    font_weight = 'bold',
    font_color = 'black',
    node_color = 'yellow',
    node_size = 800,
    font_size = 13,
    width = 3,
    pos = pos
)
```
