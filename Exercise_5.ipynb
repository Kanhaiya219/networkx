```python
import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/facebook_combined.txt')


#Number of total nodes and edges
G = nx.read_edgelist("/content/facebook_combined.txt")
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())

# first 10 nodes & edges
print(list(G.nodes())[:10])
print(list(G.edges())[:10])


```

```python
# Density of graph= 2E/N(N-1)
print("Graph Density:", nx.density(G))

#Type of graph
graph_type = "Directed" if G.is_directed() else "Undirected"
print("Graph type:", graph_type)

#Num of connected components
print("Number of connected components:", nx.number_connected_components(G))
```

```python
# Select First 50 nodes
nodes_50 = list(G.nodes())[:50]

subgraph = G.subgraph(nodes_50)

plt.figure(figsize=(12, 10))

pos = nx.spring_layout(subgraph, seed=42)

nx.draw(
    subgraph,
    pos,
    with_labels=True,
    node_size=800,
    node_color="skyblue",
    edge_color="green",
    font_size=8,
    font_weight="bold"
)
plt.title("Subgraph of 50 Nodes - Spring Layout")
plt.show()
```

```python
#Degree Analysis

# Minimum degree
degrees = dict(G.degree())
degree_series = pd.Series(degrees)

min_degree = degree_series.min()
nodes_min_degree = degree_series[degree_series == min_degree].count()

print("Minimum Degree:", min_degree)
print("Number of nodes with minimum degree:", nodes_min_degree)


#Maximum degree
max_degree = degree_series.max()
nodes_max_degree = degree_series[degree_series == max_degree].count()

print("\nMaximum Degree:", max_degree)
print("Number of nodes with maximum degree:", nodes_max_degree)


#Average degree
avg_degree = degree_series.mean()
print("\nAverage Degree:", avg_degree)


#Degree of first 10 nodes
print("\nDegree of first 10 nodes:")
first_10_nodes = list(G.nodes())[:10]
for node in first_10_nodes:
    print(f"Node {node}: Degree {degrees[node]}")

plt.figure(figsize=(10, 6))
plt.hist(degree_series, bins=30, color="skyblue", edgecolor="black")
plt.title("Degree Distribution")
plt.xlabel("Degree")
plt.ylabel("Number of Nodes")
plt.show()
```
