```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Load graph
G = nx.read_edgelist("facebook_combined.txt.gz")

# Get degree of every node
degrees = [degree for node, degree in G.degree()]

# Statistics
mean = np.mean(degrees)
median = np.median(degrees)
variance = np.var(degrees)
std_dev = np.std(degrees)

print("Degree Statistics")
print("------------------")
print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Standard Deviation:", std_dev)

#Calculate Degree Centrality and plot a line graph of it's distribution

# Calculate degree centrality
centrality = nx.degree_centrality(G)

# Sort centrality values
values = sorted(centrality.values())

# Plot distribution
plt.plot(values)

plt.xlabel("Nodes")
plt.ylabel("Degree Centrality")
plt.title("Degree Centrality Distribution")

plt.show()
```
