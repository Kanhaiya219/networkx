```python
#importing networkx
import networkx as nx
import matplotlib.pyplot as plt

G=nx.Graph()

#adding graph name and description
G.graph["Name"] = "My Graph"
G.graph["Description"] = "This Is My Graph"

#adding multiple attributes in a node
G.add_nodes_from([
    ("A",{"Age":19 , "Gender":"F"}),
    ("B",{"Age":18 , "Gender":"M"}),
    ("C",{"Age":22 , "Gender":"M"}),
    ("D",{"Age":21 , "Gender":"M"}),
    ("E",{"Age":20 , "Gender":"F"}),

])


#adding edges with customized edge thickness
G.add_edges_from([
    ("A","C",{"weight":1}),
    ("B","C",{"weight":0.5}),
    ("B","D",{"weight" : 0.6}),
    ("C","D",{"weight":0.8}),
    ("D","E",{"weight":1})
])

#adding nodes position
pos = {
    "A": (2, 4),
    "B": (5, 5.5),
    "C": (5, 2),
    "D": (6, 3),
    "E": (8, 3.5)
}


#drawing graph
nx.draw(
    G,
    with_labels=True,
    font_weight="bold",
    font_color="yellow",
    node_color="red",
    node_size=800,
    font_size=13,
    width=3,
    pos=pos
)

"""

    Print:

    • The graph attributes
    • All nodes with their attributes
    • All edges with their weights
    • Total number of nodes
    • Total number of edges
    • Degree of each node
    • Attributes of node "A"
    • Attributes of the edge between "A" and "C"
"""

print(G.graph)

for node in G.nodes(data=True):
    print(node)

for edge in G.edges(data=True):
    print(edge)

print(G.number_of_nodes())
print(G.number_of_edges())

for node in G.nodes:
    print(f"Degree({node}) = {G.degree(node)}")

print(G.nodes["A"])

print(G.edges["A", "C"])

plt.show()
```
