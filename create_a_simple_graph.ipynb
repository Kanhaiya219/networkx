```python
#importing modules
import networkx as nx
import matplotlib.pyplot as plt

#start creating a simple graph
G=nx.Graph()

#adding nodes
G.add_node("A")
G.add_node("B")
G.add_node("C")
G.add_node("D")
G.add_node("E")

#adding edges
G.add_edge("A","C")
G.add_edge("B","C")
G.add_edge("B","D")
G.add_edge("C","D")
G.add_edge("C","E")
G.add_edge("D","E")

#setting 2d coordinates (positions of nodes)
pos={
    "A" : (1,5),
    "B" : (4.5 , 6.5),
    "C" : (3.5 , 1.4),
    "D" : (5.8 , 3.5),
    "E" : (7.8 , 3.8)
}

#visualizing graph with style
nx.draw(
    G,
    with_labels=True,
    font_weight="bold",
    font_color="white",
    node_color="red",
    node_size=1000,
    font_size=15,
    width=4,
    pos=pos
)
```
